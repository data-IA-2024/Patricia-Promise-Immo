```python
import pandas as pd
from pathlib import Path
import numpy as np
```

```python
base_dir = Path().resolve().parents[1]
print(f"Base directory: {base_dir}")
interim_dir = base_dir /"Patricia-Promise-Immo"/"data" / "interim"
print(f"Interim directory: {interim_dir}")
data_no_empty_path = interim_dir / "all_immo_no_empty.csv"

df = pd.read_csv(f"{data_no_empty_path}", on_bad_lines='skip', low_memory=False, sep=',', dtype={'Code postal':str, 'Valeur fonciere':np.float64})
df
```

```python
df['Type local'] = df['Type local'].replace({
    'Dépendance': 1,
    'Maison': 2,
    'Appartement': 3
})
```

```python
# -- Filtrer seulement les ventes (si la colonne existe)
if "Nature mutation" in df.columns:
    df = df[df["Nature mutation"]=="Vente"]
df
```

```python
df.fillna(0.0, inplace=True) 
```

```python
# Calcul de la surface totale
df['surface_total']=df["Surface Carrez du 1er lot"]+df["Surface Carrez du 2eme lot"]+ df["Surface Carrez du 3eme lot"]+df["Surface Carrez du 4eme lot"]+df["Surface Carrez du 5eme lot"]
df  
```

```python
# Ne garde que les surface strictement superieur à zéro
df=df[df.surface_total>0.0] 
df
```

```python
# Calcul du prix au m2
df['prixm2']=df["Valeur fonciere"]/df.surface_total
df
```

```python
# Séparation propre les dtypes 
num_cols = df.select_dtypes(include=["number"]).columns.tolist()

num_cols
```

```python
obj_cols = df.select_dtypes(include=["object"]).columns.tolist()
obj_cols
```

```python
# Remplissage des NaN seulement sur les colonnes numériques avec 0
#df[num_cols] = df[num_cols].fillna(0)
#df
```

```python
# Construction du dataset "features" numérique en enlèvant les objets
df_final = df.drop(columns=obj_cols, errors="ignore")
df_final
```

```python
# Nettoyage éventuel de colonnes parasites
df_final= df_final.drop(columns=["Unnamed: 0","No voie","No disposition","Surface reelle bati","code_insee","Code postal","Code departement","Code commune","altitude_maximale","superficie_hectare", "B/T/Q", "5eme lot", "Prefixe de section","No plan", "latitude_centre", "longitude_centre", "Surface Carrez du 1er lot", "Surface Carrez du 2eme lot", "Surface Carrez du 3eme lot", "Surface Carrez du 4eme lot", "Surface Carrez du 5eme lot"], errors="ignore")
df_final
```

```python
# Definir le chemin de stockage du dataset final
data_dir = Path().resolve().parents[1]
data_path = data_dir /"Patricia-Promise-Immo"/"data"
processed_path = data_path/"processed" 
if not processed_path.exists():
    processed_path.mkdir(parents=True)

print(processed_path)
```
